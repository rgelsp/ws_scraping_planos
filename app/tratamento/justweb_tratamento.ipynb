# Tratamento relatório JustWeb

```python
#Biblioteca
import pandas as pd
import numpy as np

# Obtenção
df = pd.read_csv("../../data/scraped/justweb.csv")

# Código dos municípios
de_para_cod_municipio = cidades = {
                                "Belo Horizonte": 3106200,
                                "Betim": 3106705,
                                "Contagem": 3118601,
                                "Esmeraldas": 3124104,
                                "Ribeirão das Neves": 3154606,
                                "Santa Luzia": 2707909,
                                "Vespasiano": 3171204}

# Add coluna COD_MUNICIPIO
df["COD_MUNICIPIO"] = df["cidade"].map(de_para_cod_municipio)
if df["COD_MUNICIPIO"].isna().any():
    print("🟠 Há municípios não listados no de-para!")

# Add coluna de OBS
df["OBS"] = df["velocidade"] + "/" + df["valor"]

# Ajustando velocidade
df["VELOCIDADE"] = df["velocidade"].astype(str).str.extract(r"(\d+)").astype("Int64")
df["VELOCIDADE"] = np.where(df["velocidade"].str.contains("GIGA", case=False, na=False),
                              df["VELOCIDADE"]*1000,
                              df["VELOCIDADE"] )

# Add coluna Data
df["DATA"] = df["data_coleta"]
df["SVAs"] = df["svas"]
df["PRECO"] = df["valor"]
df["PERIODO"] = pd.NA
df["PRECO_DEPOIS"] = pd.NA

# Renomear coluna
df = df.rename(columns={"provedor": "ISP"})

df["ID"] = df["ISP"].astype(str) + "_" + df["DATA"] + "_" + (df.index + 1).astype(str)

df["OBS"] = df["velocidade"] + "/" + df["cidade"]

colunas = [
    "ID", "DATA", "ISP", "VELOCIDADE", "SVAs",
    "PRECO", "PERIODO", "PRECO_DEPOIS", "COD_MUNICIPIO"
]

df = df[colunas]

# Salvando
nome_arq = df["DATA"].max() + "-JUSTWEB.csv"
pasta = r"\\fileserver\arquivos\Estrategia Comercial\2. POWER BI\ECM - Radar Mercado\ScrapingPlanos"
df.to_csv(fr"{pasta}\{nome_arq}", index=False)
print(f"🟢 {nome_arq} salvo!")

df
```
