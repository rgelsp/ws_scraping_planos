# Tratamento Relatório Ibe

```python
import pandas as pd
import numpy as np
import re

# Leitura
df = pd.read_csv("../../data/scraped/ibe.csv")

# =========================
# PADRONIZAÇÃO CAMPOS BASE
# =========================

# ISP
df["ISP"] = "Ibitelecom"

# DATA
df["DATA"] = df["data_coleta"]

df["VELOCIDADE"] = (
    df["nome_plano"]
    .astype(str)
    .str.extract(r"(\d+)\s*(?=mega|mb)", flags=re.IGNORECASE)[0]
    .astype("Int64")
)

# PRECO
df["PRECO"] = (
    df["valor"]
    .astype(str)
    .str.replace(",", ".", regex=False)
    .str.extract(r"(\d+\.?\d*)")[0]
    .astype(float)
)


# SVAs
df["SVAs"] = (
    df["nome_plano"]
    .astype(str)
    .str.split("+", n=1)
    .str[1]
    .where(lambda x: x.notna(), pd.NA)
    .str.strip()
)

# Campos padrão
df["PERIODO"] = pd.NA
df["PRECO_DEPOIS"] = pd.NA

# =========================
# EXPANSÃO MUNICÍPIOS
# =========================

de_para_cod_municipio = {
    "Governador Valadares": 3127701,
    "Ipatinga": 3131307
}

# cria lista de códigos para cada linha
df["COD_MUNICIPIO"] = [list(de_para_cod_municipio.values())] * len(df)

# explode (replica planos)
df = df.explode("COD_MUNICIPIO").reset_index(drop=True)

# =========================
# ID
# =========================

df["ID"] = (
    df["ISP"].astype(str)
    + "_"
    + df["DATA"].astype(str)
    + "_"
    + (df.index + 1).astype(str)
)

# =========================
# COLUNAS FINAIS
# =========================

colunas = [
    "ID", "DATA", "ISP", "VELOCIDADE", "SVAs",
    "PRECO", "PERIODO", "PRECO_DEPOIS", "COD_MUNICIPIO"
]

df = df[colunas]

# =========================
# EXPORT
# =========================

nome_arq = df["DATA"].max() + "-IBE.csv"
pasta = r"\\fileserver\arquivos\Estrategia Comercial\2. POWER BI\ECM - Radar Mercado\ScrapingPlanos"

df.to_csv(fr"{pasta}\{nome_arq}", index=False)

print(f"🟢 {nome_arq} salvo!")

df
```
