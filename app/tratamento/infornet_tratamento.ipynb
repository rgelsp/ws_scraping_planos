# Tratamento relatório Infornet

```python
import pandas as pd
import numpy as np

# Obtenção
df = pd.read_csv("../../data/scraped/infornet.csv")

# De-para municípios (ajuste se necessário)
de_para_cod_municipio = {
    "João Monlevade": 3136207,
    "Ponte Nova": 3152105,
    "Rio Piracicaba": 3155702
}

# COD_MUNICIPIO
df["COD_MUNICIPIO"] = df["cidade"].map(de_para_cod_municipio)
if df["COD_MUNICIPIO"].isna().any():
    print("🟠 Municípios não mapeados!")

# VELOCIDADE (já numérica, padronizar)
df["VELOCIDADE"] = pd.to_numeric(df["velocidade"], errors="coerce")
df["VELOCIDADE"] = df["VELOCIDADE"].apply(
    lambda x: x * 1000 if pd.notna(x) and x < 10 else x
)
df["VELOCIDADE"] = df["VELOCIDADE"].round().astype("Int64")

# PRECO (corrigir vírgula → float)
df["PRECO"] = (
    df["valor"]
    .astype(str)
    .str.replace(",", ".", regex=False)
    .astype(float)
)


# DATA
df["DATA"] = df["data_coleta"]

# SVAs
df["SVAs"] = df["svas"]

# Campos padrão
df["PERIODO"] = pd.NA
df["PRECO_DEPOIS"] = pd.NA

# ISP
df = df.rename(columns={"provedor": "ISP"})

# ID
df["ID"] = df["ISP"].astype(str) + "_" + df["DATA"] + "_" + (df.index + 1).astype(str)

# Colunas finais
colunas = [
    "ID", "DATA", "ISP", "VELOCIDADE", "SVAs",
    "PRECO", "PERIODO", "PRECO_DEPOIS", "COD_MUNICIPIO"
]

df = df[colunas]

# Salvando
nome_arq = df["DATA"].max() + "-INFORNET.csv"
pasta = r"\\fileserver\arquivos\Estrategia Comercial\2. POWER BI\ECM - Radar Mercado\ScrapingPlanos"

df.to_csv(fr"{pasta}\{nome_arq}", index=False)
print(f"🟢 {nome_arq} salvo!")

df
```
