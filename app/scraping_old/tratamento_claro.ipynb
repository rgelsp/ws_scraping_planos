# Tratamento relatório Claro

```python
#Biblioteca
import pandas as pd

# Obtenção
df = pd.read_csv("../../data/scraped/claro.csv")

# ID
df["ID"] = "Claro_" + df["data_coleta"] + "_" + (df.index + 1).astype(str)

# PROVEDOR, SVAs, DATA
df = df.rename(columns={"provedor": "ISP",
                        "svas": "SVAs",
                        "data_coleta": "DATA"})

# VELOCIDADE
df["VELOCIDADE"] = df["velocidade"].astype(str).str.extract(r"(\d+)").astype("Int64")

# PRECO
df["PRECO"] = df["valor"].str.split("$").str[1].str.replace(",", ".", regex=False).astype(float)

# PERIODO
df["PERIODO"] = df["periodo_promo"].astype(str).str.extract(r"(\d+)").astype("Int64")

# PRECO_DEPOIS
df["PRECO_DEPOIS"] = df["vr_pos_promo"].str.split("$").str[1]\
    .str.split("/").str[0]\
    .str.replace(",", ".", regex=False)

# COD_MUNICIPIO
de_para_cod_municipio = {"belo-horizonte":3106200,
                         "contagem":3118601,
                         "governador-valadares":3127701,
                         "ipatinga":3131307,
                         "itabira":3131703,
                         "ribeirao-das-neves":3154606,
                         "sabara":3156700,
                         "santa-luzia":3157807,
                         "vespasiano":3171204,
                         "ouro preto":3146107,
                         "mariana":3140001,
                         "betim":3106705,
                         "ponte-nova":3152105,
                         "joao-monlevade":3136207}

# Add coluna COD_MUNICIPIO
df["COD_MUNICIPIO"] = df["cidade"].map(de_para_cod_municipio)
if df["COD_MUNICIPIO"].isna().any():
    print("🟠 Há municípios não listados no de-para!")

# Add coluna OBS
df['OBS'] = df['velocidade'] + '/' + df["cidade"]


df = df[["ID", "DATA", "ISP", "VELOCIDADE", "SVAs", "PRECO", "PERIODO", "PRECO_DEPOIS", "COD_MUNICIPIO", "OBS"]]

# Salvando
nome_arq = df["DATA"].max() + "-CLARO.csv"
pasta = r"\\fileserver\arquivos\Estrategia Comercial\2. POWER BI\ECM - Radar Mercado\ScrapingPlanosMunicipios"
df.to_csv(fr"{pasta}\{nome_arq}", index=False)
print(f"🟢 {nome_arq} salvo!")

df
```
