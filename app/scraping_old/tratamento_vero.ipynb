# Tratamento relatório Vero

```python
#Biblioteca
import pandas as pd

# Obtenção
df = pd.read_csv("../../data/scraped/vero.csv")

# Código dos municípios
de_para_cod_municipio = {"Belo Horizonte - MG": 3106200,
                        "Betim - MG": 3106705,
                        "Contagem - MG": 3118601,
                        "Esmeraldas - MG": 3124104,
                        "Governador Valadares - MG": 3127701,
                        "Itabira - MG": 3131703,
                        "Itabirito - MG": 3131901,
                        "João Monlevade - MG": 3136207,
                        "Mariana - MG": 3140001,
                        "Ouro Branco - MG": 3145901,
                        "Ouro Preto - MG": 3146107,
                        "Ponte Nova - MG": 3152105,
                        "Ribeirão das Neves - MG": 3154606,
                        "Sabará - MG": 3156700,
                        "Santa Bárbara - MG": 2927507,
                        "Santa Luzia - MG": 3157807,
                        "Vespasiano - MG": 3171204
                         }

# Add coluna COD_MUNICIPIO
df["COD_MUNICIPIO"] = df["MUNICIPIO"].map(de_para_cod_municipio)
if df["COD_MUNICIPIO"].isna().any():
    print("🟠 Há municípios não listados no de-para!")

# Add coluna de OBS
df["OBS"] = df["TITULO"] + "/" + df["MUNICIPIO"]

# Ajustando velocidade (pega apenas números)
df["VELOCIDADE"] = df["VELOCIDADE"].astype(str).str.extract(r"(\d+)").astype("Int64")

# Add coluna de ID
df["ISP"] = "Vero"
df["ID"] = df["ISP"].astype(str) + "_" + df["DATA"] + "_" + (df.index + 1).astype(str)

# Ajuste Celular
df["CELULAR"] = ", " + df["TITULO"].str.extract(r"(Celular \d+GB)")
df["CELULAR"] = df["CELULAR"].fillna("")  # evita NaN

# Ajuste SVAs
df["SVAs"] = df["SVAs"].str.replace(",", ", ", regex=False)
# Add Celular aos SVAs
df["SVAs"] = df["SVAs"] + df["CELULAR"]

# Add coluna adiconais
df["PERIODO"] = pd.NA
df["PRECO_DEPOIS"] = pd.NA

df = df[["ID", "DATA", "ISP", "VELOCIDADE", "SVAs", "PRECO", "PERIODO", "PRECO_DEPOIS", "COD_MUNICIPIO", "OBS"]]

# Salvando
nome_arq = df["DATA"].max() + "-VERO.csv"
pasta = r"\\fileserver\arquivos\Estrategia Comercial\2. POWER BI\ECM - Radar Mercado\ScrapingPlanosMunicipios"
df.to_csv(fr"{pasta}\{nome_arq}", index=False)
print(f"🟢 {nome_arq} salvo!")

df
```
